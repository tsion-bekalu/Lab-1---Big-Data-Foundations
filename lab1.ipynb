#### **Task 5(stretch) - Conditional Probability, Revisited** 

```python
p_default = (df["defaulted"] == "Yes").mean()
print("P(Default):", p_default)

rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)
print(rate_by_collateral)
```

<div style="width: 98%; max-width: 100%; box-sizing: border-box; border: 1px solid #ccc; border-radius: 10px; padding: 15px; overflow-wrap: break-word;">

##### <u>Answers</u>

- The overall probability of an applicant defaulting is:
  - P(Default) = 0.1924

- The probability of default for applicants who have collateral is:
  - P(Default | Has Collateral) = 0.1208

- The probability of default for applicants who do not have collateral is:
  - P(Default | No Collateral) = 0.26


- Yes. Applicants without collateral default more often.
  - 0.26 / 0.1208 = 2.15

  This shows applicants without collateral defaults 2.15 times as often. The same pattern from Unit 1 hold here.
